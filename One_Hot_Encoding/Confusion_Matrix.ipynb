```python
# 3. Classify a person as male or female based on a given dataset using naïve
# Bayesian   Classifier, and calculate accuracy, precision, and recall for your
# data set

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

# 1. Create a synthetic dataset (replace with your actual dataset)
data = {
    'height_cm': [170, 165, 180, 160, 175, 155, 185, 168],
    'weight_kg': [70, 55, 85, 50, 78, 48, 90, 60],
    'foot_size_cm': [28, 24, 30, 23, 29, 22, 31, 25],
    'gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female']
}
df = pd.DataFrame(data)

# Convert 'gender' to numerical labels
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})

# Separate features (X) and target (y)
X = df[['height_cm', 'weight_kg', 'foot_size_cm']]
y = df['gender']
# 2. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Initialize and train the Naive Bayes Classifier (Gaussian Naive Bayes for
# continuous data)
model = GaussianNB()
model.fit(X_train, y_train)

# 4. Make predictions on the test set
y_pred = model.predict(X_test)

# 5. Calculate and print evaluation metrics
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print(f"Precision (Male): {precision_score(y_test, y_pred, pos_label=1):.2f}")
print(f"Recall (Male): {recall_score(y_test, y_pred, pos_label=1):.2f}")

# 6 For a more detailed report including both classes
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['Female', 'Male']))

# 7 You can also make a prediction on a new, unseen sample
new_sample = [[160, 48, 21],[180, 55, 30],[140, 46, 28]]
 # Example new sample
predicted_class = model.predict(new_sample)
print(f"Predicted class for new sample: {predicted_class:}")
```

```python
import pandas as pd

# Original dataset
data = {
    'NAME': ['MOHAN', 'SHYAM', 'RAVI', 'RAM'],
    'SPORTS': ['HOCKEY', 'CRICKET', 'BADMINTON', 'CHESS'],
    'LABEL': ['BEGINNER', 'INTERMEDIATE', 'ADVANCED', None]  # RAM has no label
}

# Create DataFrame
df = pd.DataFrame(data)

# Add Indoor/Outdoor column
indoor_sports = ['BADMINTON', 'CHESS']
df['INDOOR'] = df['SPORTS'].apply(lambda x: 1 if x in indoor_sports else 0)

# Apply One Hot Encoding on 'SPORTS'
df_encoded = pd.get_dummies(df, columns=['SPORTS'])

print("Final Dataset:\n", df_encoded)
```
