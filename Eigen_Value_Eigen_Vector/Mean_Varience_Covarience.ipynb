```python
import numpy as np
import pandas as pd

# Dataset
x = np.array([2, 0, 1, 3], dtype=float)
y = np.array([0, 2, 1, 3], dtype=float)

# 1. Mean
mean_x = np.mean(x)
mean_y = np.mean(y)

print("Mean of x:", mean_x)
print("Mean of y:", mean_y)

# 2. Variance (sample variance, ddof=1)
var_x = np.var(x, ddof=1)
var_y = np.var(y, ddof=1)

print("\nSample Variance of x:", var_x)
print("Sample Variance of y:", var_y)

# 3. Covariance
cov_matrix = np.cov(x, y, ddof=1)  # sample covariance matrix
print("\nCovariance Matrix:\n", cov_matrix)


# 4. Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print ("\nEigenvalues:", eigenvalues)
print("Eigenvectors (columns correspond to eigenvalues):\n", eigenvectors)
```
